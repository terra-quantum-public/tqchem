### Calculate and plot the IR spectrum

Now we calculate IR excitation frequencies and corresponding intensities.
These individual excitations are plotted as a spectrum by first broadening them using Lorentzians with a given width.

```python
from tqchem.xtb_api import xtb_ir_spectrum
from tqchem.spectra import plot_spectrum


frequencies, intensities = xtb_ir_spectrum(optimized, method="gfn1-xtb")

fig = plot_spectrum(frequencies, intensities, broaden_type="lorentzian")
fig.show()
```
